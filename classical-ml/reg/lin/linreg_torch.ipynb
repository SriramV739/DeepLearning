```python
import torch 
import pandas as pd
import numpy as np
import torch.nn.functional as F
```

```python
df=pd.read_csv("data.csv")
num_rows=df.shape[0]
num_cols=df.shape[1]-1
print(num_rows,num_cols)
```

```python
train_data=df.iloc[:int(0.8*num_rows),:-1]
train_labels=df.iloc[:int(0.8*num_rows),-1]
train_data_tensor=torch.tensor(train_data.values,dtype=torch.float32)
train_label_tensor=torch.tensor(train_labels.values,dtype=torch.float32)


train_mean = train_data_tensor.mean(dim=0)
train_std = train_data_tensor.std(dim=0)

train_data_tensor_norm = (train_data_tensor - train_mean) / train_std
train_label_tensor_norm = (train_label_tensor - train_label_tensor.mean()) / train_label_tensor.std()


```

```python
epochs=10000
lr=0.01

W=torch.randn(num_cols,1,requires_grad=True)
while epochs:
    epochs-=1
    
    values=(train_data_tensor_norm @ W).squeeze(1)
    loss=F.mse_loss(values,train_label_tensor_norm)
    print(loss)
    loss.backward()
    with torch.no_grad():
        W-=lr*W.grad
    
    W.grad=None

```

```python
test_data=df.iloc[int(0.8*num_rows):,:-1]
test_labels=df.iloc[int(0.8*num_rows):,-1]
test_data_tensor=torch.tensor(test_data.values,dtype=torch.float32)
test_label_tensor=torch.tensor(test_labels.values,dtype=torch.float32)


test_mean = test_data_tensor.mean(dim=0)
test_std = test_data_tensor.std(dim=0)

test_data_tensor_norm = (test_data_tensor - test_mean) / test_std
test_label_tensor_norm = (test_label_tensor - test_label_tensor.mean()) / test_label_tensor.std()

```

```python
values=(test_data_tensor_norm @ W).squeeze(1)
loss=F.mse_loss(values,test_label_tensor_norm)
print(loss)
```
